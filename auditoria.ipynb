# Auditoría del conjunto de datos

```python
import pandas as pd
```

```python
df_pedidos = pd.read_csv('data/raw/DW_Pedidos.csv')

print(df_pedidos.shape)
```

## Tipos de datos y valores faltantes

```python
df_resumen = pd.DataFrame({
    "Tipo de dato": df_pedidos.dtypes,
    "Valores no nulos": df_pedidos.notnull().sum(),
    "Valores faltantes": df_pedidos.isnull().sum()
})

print(df_resumen)
```

## Valores únicos por columna 
 
- Analizamos las columnas que son Strings para buscar casos en los que hayan escrito lo mismo de formas distintas

```python
# Valores únicos por columna (ordenados alfabéticamente)
for columna in df_pedidos:
    print(f"\n{columna}:")
    # sorted() ordena los valores únicos de forma ascendente
    valores_ordenados = sorted(df_pedidos[columna].dropna().unique())
    print(valores_ordenados)
```

## Conversión de los tipos de datos de las columnas 

- Convertir los objects al tipo de dato correcto "string".

Ayuda a que en el tablero consuma menos espacio al importar los datos y permita utilizar ciertas características propias del tipo de dato. Por ejemplo, ordenar alfabéticamente las variables de tipo texto.


```python
fecha_minima = df_pedidos["FechaPedido"].min()
fecha_maxima = df_pedidos["FechaPedido"].max()
print(f"\nRango de fechas: {fecha_minima} a {fecha_maxima}")
```

```python
cols_fecha = ['FechaNacimiento', 'FechaContratación', 'FechaPedido', 'FechaEntrega', 'FechaEnvío']
for col in cols_fecha:
    df_pedidos[col] = pd.to_datetime(df_pedidos[col], errors='coerce')

cols_texto = df_pedidos.select_dtypes(include=['object']).columns

for col in cols_texto:
    df_pedidos[col] = df_pedidos[col].astype('string')


print(df_pedidos.dtypes)
```
